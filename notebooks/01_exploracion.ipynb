```python
import pandas as pd
pd.set_option("display.max_columns", None)

df = pd.read_csv("../data/raw/04OCLicitacion.csv", sep=";", encoding="latin-1",
                 decimal=",", low_memory=False,
                 dtype={"codigoOC": str, "ProveedorRUT": str, "UnidadCompraRUT": str})

print(df[["MontoNetoOC", "MontoNetoOC_CLP", "MontoNetoItemCLP"]].dtypes)
```

```python
montos = ["MontoNetoOC_CLP", "MontoNetoItemCLP"]
print("Nulos:", df[montos].isna().sum().to_dict())
print("Ceros o negativos:", (df[montos] <= 0).sum().to_dict())
print("Filas duplicadas:", df.duplicated().sum())
print(df["MontoNetoItemCLP"].describe().round(0))

# ¿La suma de los ítems calza con el monto de la orden?
chk = df.groupby("codigoOC").agg(suma_items=("MontoNetoItemCLP", "sum"),
                                 monto_oc=("MontoNetoOC_CLP", "first"))
chk["dif_pct"] = (chk.suma_items - chk.monto_oc).abs() / chk.monto_oc * 100
print("Órdenes que calzan (<1% de diferencia):", round((chk.dif_pct < 1).mean() * 100, 1), "%")
print(chk.sort_values("dif_pct", ascending=False).head(5))
```

```python
cols = ["codigoOC", "Institucion", "Proveedor", "NombreItem", "MonedaOC", "CantidadItem", "MontoNetoItemCLP"]
top = df.nlargest(10, "MontoNetoItemCLP")[cols].copy()
top["Institucion"] = top["Institucion"].str[:28]
top["Proveedor"] = top["Proveedor"].str[:25]
top["NombreItem"] = top["NombreItem"].str[:35]
print(top.to_string())

total = df["MontoNetoItemCLP"].sum()
print("\nGasto total (millones de CLP):", round(total / 1e6))
print("Los 10 ítems más grandes son el", round(top["MontoNetoItemCLP"].sum() / total * 100, 1), "% del total")
print(df["MontoNetoItemCLP"].quantile([0.5, 0.9, 0.99, 0.999]).round(0))
```

```python
dup = df[df.duplicated(keep=False)]
print("Filas duplicadas (todas las copias):", len(dup), "| Órdenes afectadas:", dup["codigoOC"].nunique())
print(dup.sort_values("codigoOC")[["codigoOC", "NombreItem", "CantidadItem", "MontoNetoItemCLP"]].head(8).to_string())

print("\nMontos menores a $1:", (df["MontoNetoItemCLP"] < 1).sum())
print("\nFilas con monto nulo:", df["MontoNetoItemCLP"].isna().sum(), "| Estados:")
print(df[df["MontoNetoItemCLP"].isna()]["EstadoOC"].value_counts())
```

```python
import pandas as pd, sqlite3

archivos = {
    "Licitacion": "../data/raw/04OCLicitacion.csv",
    "Convenio Marco": "../data/raw/04OCConvenioMarco.csv",
    "Trato Directo": "../data/raw/04OCTratoDirecto.csv",
}
cols = ["codigoOC", "FechaEnvioOC", "EstadoOC", "MonedaOC", "MontoNetoOC_CLP",
        "Institucion", "UnidadCompra", "RegionUnidadCompra", "Sector",
        "Proveedor", "ProveedorRUT", "TamanoProveedor", "RegionProveedor",
        "RubroN1", "RubroN2", "RubroN3", "NombreItem", "CantidadItem",
        "UnidadMedida", "MontoNetoItemCLP"]

partes = []
for modalidad, ruta in archivos.items():
    d = pd.read_csv(ruta, sep=";", encoding="latin-1", decimal=",", usecols=cols,
                    low_memory=False, dtype={"codigoOC": str, "ProveedorRUT": str})
    d["Modalidad"] = modalidad
    partes.append(d)
df = pd.concat(partes, ignore_index=True)

# Fechas en formato ISO (SQLite no tiene tipo fecha)
f = pd.to_datetime(df["FechaEnvioOC"], format="%d-%m-%Y %H:%M:%S", errors="coerce")
df["Fecha"] = f.dt.strftime("%Y-%m-%d")
df["Mes"] = f.dt.strftime("%Y-%m")
df = df.drop(columns=["FechaEnvioOC"])

# Banderas según docs/decisiones.md
df["EsMontoValido"] = df["EstadoOC"].isin(["Aceptada", "Recepcion Conforme"]).astype(int)
df["EsDatoInvalido"] = (df["codigoOC"] == "4707-160-SE25").astype(int)
df["EsContratoGrande"] = (df["MontoNetoItemCLP"] >= 50_000_000).astype(int)
df["RegionUnidadCompra"] = df["RegionUnidadCompra"].fillna("Sin información")
df["RegionProveedor"] = df["RegionProveedor"].fillna("Sin información")

con = sqlite3.connect("../data/chilecompra.db")
df.to_sql("ordenes_items", con, if_exists="replace", index=False)
print("Tabla guardada:", len(df), "filas")
```

```python
q = """
SELECT Modalidad,
       COUNT(DISTINCT codigoOC)  AS ordenes,
       ROUND(SUM(MontoNetoItemCLP) / 1e6)  AS monto_millones,
       SUM(EsContratoGrande)  AS items_contrato_grande
FROM ordenes_items
WHERE EsMontoValido = 1 AND EsDatoInvalido = 0
GROUP BY Modalidad
"""
print(pd.read_sql(q, con))
print(pd.read_sql("SELECT COUNT(*) AS filas_totales FROM ordenes_items", con))
```
